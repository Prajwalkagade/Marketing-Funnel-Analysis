```python
import pandas as pd

# Load dataset
df = pd.read_csv("../data/raw/marketing_funnel_raw.csv", sep=';')

# Show first 5 rows
df.head()
```

```python
# Dataset information
df.info()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Check duplicate rows
df.duplicated().sum()
```

```python
# Convert target column into numeric
df['y'] = df['y'].map({'yes': 1, 'no': 0})

# Check conversion
df['y'].value_counts()
```

```python
# Create funnel stages

df['Visitors'] = 1

df['Leads'] = df['duration'].apply(lambda x: 1 if x > 100 else 0)

df['Customers'] = df['y']

# Preview
df[['Visitors', 'Leads', 'Customers']].head()
```

```python
# Conversion metrics

df['Lead_Conversion_Rate'] = (
    df['Leads'] / df['Visitors']
) * 100

df['Customer_Conversion_Rate'] = (
    df['Customers'] / df['Visitors']
) * 100

# Preview
df[['Lead_Conversion_Rate',
    'Customer_Conversion_Rate']].head()
```

```python
# Save cleaned dataset

df.to_csv(
    "../data/cleaned/marketing_funnel_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
